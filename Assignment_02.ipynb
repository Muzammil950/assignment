```python
def odd_nums():
    odd_list = []
    for i in range(0, 26):
        if i % 2 != 0:
            odd_list.append(i)
    return odd_list       
```

```python
odd_nums()
```

```python
def generate_squares(n):
    for i in range(1,n+1):
        yield i*i
```

```python
squares = generate_squares(5)
```

```python
for value in squares:
    print(value)
```
