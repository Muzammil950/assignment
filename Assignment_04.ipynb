```python
class vehicle :
    def __init__(self, name_vehicle, max_speed, average_of_vehicle):
        self.name_vehicle = name_vehicle
        self.max_speed = max_speed
        self.average_of_vehicle = average_of_vehicle

    def vehicle_details(self):
        return self.max_speed, self.name_vehicle, self.average_of_vehicle
        
```

```python
rohan = vehicle("suzuki", 20, 35)
```

```python
rohan.vehicle_details()
```

```python
class vehicle1 :
    def __init__(self, name, max_speed, mileage):
        self.name = name
        self.max_speed = max_speed
        self.mileage = mileage

        
```

```python
class car(vehicle1):
    def seating_capacity(self, capacity):
        return f"the seating capacity of {self.name} is {capacity} passenger"
        
```

```python
car1 = car("innova", 180, 15)
```

```python
car1.seating_capacity(7)
```

```python
class engine:
    def engine_info(self):
        return "this is a power ful engine"

class wheels:
    def wheels_info(self):
        return "this car has 4 alloy wheels"

class car(engine, wheels):
    def car_info(self):
        return "this ia a modern car"
        
```
