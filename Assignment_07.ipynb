#Write a Python program to create a Student class with attributes like name, age, and marks. Add a method to display the student’s details.

```python
class student :
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def student_info(self):
        return self.name, self.age, self.marks
```

```python
s1 = student("asif", 24, 80)
s2 = student("pandu", 22, 100)
```

```python
s1.student_info()
```

```python
s2.student_info()
```

#Create a Car class with attributes brand, model, and mileage. Add a method to calculate and print fuel used if the car travels a certain distance.

```python
class car : 
    def __init__(self, brand, model, mileage):
        self.brand = brand
        self.model = model
        self.mileage = mileage

    def fuel_used(self, distance):
        fuel = distance/self.mileage
        print(f"{self.brand} {self.model} used {fuel:.2f} for {distance} km.")
```

```python
car1 = car("toyota","fortuner",12)
```

```python
car1.fuel_used(240)
```

#Create a class `Employee` with attributes `name`, `salary`. Add a method `calculate_bonus()` that gives a 10% bonus if salary < 50000, else 5%. Print the final salary after adding the bonus.

```python
class Employee:
    
    def __init__(self, name, salary):
        self.name = name          
        self.salary = salary      

   
    def calculate_bonus(self):
        if self.salary < 50000:
            bonus = self.salary * 0.10   
        else:
            bonus = self.salary * 0.05   

        final_salary = self.salary + bonus
        print(f"Employee: {self.name}")
        print(f"Base Salary: ₹{self.salary}")
        print(f"Bonus: ₹{bonus:.2f}")
        print(f"Final Salary after Bonus: ₹{final_salary:.2f}")
        print("-" * 40)
```

```python
emp1 = Employee("Rahul", 40000)
emp2 = Employee("Neha", 60000)
```

```python
emp1.calculate_bonus()
emp2.calculate_bonus()
```

#Create a `BankAccount` class with methods: * `deposit(amount)` * `withdraw(amount)` * `display_balance()` Start with a balance of ₹0 and update it after every operation.

```python
class BankAccount:
    
    def __init__(self):
        self.balance = 0

    
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited: ₹{amount}")
    
   
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance!")
        else:
            self.balance -= amount
            print(f"Withdrawn: ₹{amount}")
    
    
    def display_balance(self):
        print(f"Current Balance: ₹{self.balance}")
```

```python
my_account = BankAccount()
```

```python
my_account.deposit(5000)
my_account.display_balance()

my_account.withdraw(1500)
my_account.display_balance()

my_account.withdraw(4000)  
my_account.display_balance()
```
